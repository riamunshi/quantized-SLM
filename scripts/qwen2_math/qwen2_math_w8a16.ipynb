```python
from datasets import load_dataset, concatenate_datasets
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig
import torch
import random
import time
import re
import pandas as pd
```

```python
SEED = 42
random.seed(SEED)
```

```python
subjects = ["algebra", "geometry", "number_theory", "prealgebra"]

datasets_list = []

for sub in subjects:
    ds_part = load_dataset("EleutherAI/hendrycks_math", sub, split="train")
    datasets_list.append(ds_part)

ds = concatenate_datasets(datasets_list)

print("Total samples:", len(ds))
```

```python
easy, medium, hard = [], [], []

for i, row in enumerate(ds):
    level = row['level']

    if "1" in level:
        easy.append(i)
    elif "3" in level:
        medium.append(i)
    elif "5" in level:
        hard.append(i)

random.shuffle(easy)
random.shuffle(medium)
random.shuffle(hard)

# 🔥 FULL 50 SAMPLE
sampled = easy[:20] + medium[:20] + hard[:10]

problems = [ds[i] for i in sampled]

print("Sampled problems:", len(problems))
```

```python
model_name = "Qwen/Qwen2-1.5B"

tokenizer = AutoTokenizer.from_pretrained(model_name)

bnb_config = BitsAndBytesConfig(load_in_8bit=True)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    quantization_config=bnb_config,
    device_map="auto"
)

device = "cuda" if torch.cuda.is_available() else "cpu"

print("W8A16 model loaded!")
```

```python
def solve(problem):
    prompt = f"Question: {problem['problem']}\nAnswer:"

    inputs = tokenizer(prompt, return_tensors="pt").to(device)

    outputs = model.generate(
        **inputs,
        max_new_tokens=200,
        do_sample=False
    )

    return tokenizer.decode(outputs[0], skip_special_tokens=True)
```

```python
def extract_number(text):
    nums = re.findall(r'-?\d+', text)
    return int(nums[-1]) if nums else None
```

```python
print("="*60)
print("Strategy: greedy | Quant: W8A16 | 50 samples")
print("="*60)

results = []
start_time = time.time()

for i, p in enumerate(problems):
    output = solve(p)

    pred = extract_number(output)
    gt = extract_number(p["solution"])

    correct = (pred == gt)

    elapsed = int((time.time() - start_time) / 60)

    print(f"[{i+1}/{len(problems)}] "
          f"diff={p['level']} "
          f"pred={pred} gt={gt} "
          f"correct={correct} "
          f"elapsed={elapsed}min")

    results.append({
        "pid": i,
        "difficulty": p["level"],
        "prediction": pred,
        "ground_truth": gt,
        "correct": correct,
        "model_output": output
    })
```

```python
df = pd.DataFrame(results)
df.to_csv("qwen2_math_w8a16_50.csv", index=False)

print("CSV saved!")
```
